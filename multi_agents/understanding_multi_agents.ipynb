```python
print("=" * 60)
print("STEP 1: Importing Required Libraries")
print("=" * 60)
print("Importing autogen_agentchat.agents - For creating AI agents")
print("Importing autogen_ext.models.openai - For OpenAI model client")
print("Importing pydantic - For defining structured data models")
print("Importing dotenv - For loading environment variables")
print("Importing os - For accessing environment variables")
print("-" * 60)

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from pydantic import BaseModel, Field
from dotenv import load_dotenv
from autogen_agentchat.messages import TextMessage
import os

print("✓ All libraries imported successfully!\n")
```

```python
print("=" * 60)
print("STEP 2: Loading Environment Variables")
print("=" * 60)
print("Loading .env file to access API keys securely...")

load_dotenv()  # Load variables from .env file
openai_key = os.getenv("OPEN_AI_KEY")

if openai_key:
    print(f"✓ API key loaded successfully (Length: {len(openai_key)} characters)")
else:
    print("❌ Warning: API key not found!")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 3: Creating Multi-Agent Team")
print("=" * 60)
print("Setting up two agents that will work together...")
print("-" * 60 + "\n")

# Import team components
from autogen_agentchat.teams import RoundRobinGroupChat, MaxMessageTermination

# Create model client
model_client = OpenAIChatCompletionClient(
    model="gpt-4o",
    api_key=openai_key
)

print("Creating Agent 1 and Agent 2...")
agent1 = AssistantAgent(
    "Assistant1", 
    model_client=model_client,
    system_message="You are Assistant 1. Count numbers briefly."
)
agent2 = AssistantAgent(
    "Assistant2", 
    model_client=model_client,
    system_message="You are Assistant 2. Count numbers briefly."
)

print("✓ Two agents created!")
print("-" * 60 + "\n")

print("Setting up Round Robin team (max 6 messages)...")
termination = MaxMessageTermination(6)
team = RoundRobinGroupChat([agent1, agent2], termination_condition=termination)

print("✓ Team created!")
print("  Type: Round Robin (agents take turns)")
print("  Termination: After 6 messages")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 4: Running the Team")
print("=" * 60)
print("Task: Count from 1 to 10, one number at a time")
print("-" * 60 + "\n")

# Run the team - Use 'await' for async execution
result = await team.run(task="Count from 1 to 10, respond one at a time.")

print("\n" + "=" * 60)
print("RESULT:")
print("=" * 60)
print(result)
```
