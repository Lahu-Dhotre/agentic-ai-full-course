```python
print("=" * 60)
print("STEP 1: Importing Required Libraries")
print("=" * 60)
print("Importing autogen_agentchat.agents - For creating AI agents")
print("Importing autogen_ext.models.openai - For OpenAI model client")
print("Importing pydantic - For defining structured data models")
print("Importing dotenv - For loading environment variables")
print("Importing os - For accessing environment variables")
print("-" * 60)

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from pydantic import BaseModel, Field
from dotenv import load_dotenv
from autogen_agentchat.messages import TextMessage
import os

print("✓ All libraries imported successfully!\n")
```

```python
print("=" * 60)
print("STEP 2: Loading Environment Variables")
print("=" * 60)
print("Loading .env file to access API keys securely...")

load_dotenv()  # Load variables from .env file
openai_key = os.getenv("OPEN_AI_KEY")

if openai_key:
    print(f"✓ API key loaded successfully (Length: {len(openai_key)} characters)")
else:
    print("❌ Warning: API key not found!")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 3: Defining Structured Output Schema with Pydantic")
print("=" * 60)
print("Creating a PlanetInfo class to define the structure of data")
print("This ensures the AI returns data in a consistent format")
print("-" * 60)

class PlanetInfo(BaseModel):
    name: str = Field(..., description="The name of the planet")
    distance_from_sun_km: float = Field(..., description="Distance from the sun in kilometers")
    has_life: bool = Field(..., description="Indicates if the planet supports life")

print("✓ PlanetInfo schema defined with 3 fields:")
print("  1. name (string) - Planet name")
print("  2. distance_from_sun_km (float) - Distance in km")
print("  3. has_life (boolean) - Life support indicator")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 4: Creating Model Client (The AI Brain)")
print("=" * 60)
print("Configuring OpenAI model with structured output...")
print("Using gpt-4o (supports structured output)")
print("Setting response_format=PlanetInfo to enforce structured responses")
print("-" * 60)

model_client = OpenAIChatCompletionClient(
    model="gpt-4o",  # FIXED: Changed from gpt-4 to gpt-4o (supports structured output)
    api_key=openai_key,
    response_format=PlanetInfo  # KEY: This enforces structured output!
)

print("✓ Model client created successfully!")
print("  Model: gpt-4o (supports structured output)")
print("  Response Format: PlanetInfo (structured)")
print("  The AI will ONLY return data matching our PlanetInfo schema")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 5: Creating the AI Agent")
print("=" * 60)
print("Initializing AssistantAgent with:")
print("  - Name: 'assistant'")
print("  - Model: The GPT-4 client we configured")
print("  - System Message: Instructions for the agent")
print("-" * 60)

agent = AssistantAgent(
    name="assistant",
    model_client=model_client,
    system_message="You're an AI assistant that provides structured information about planets."
)

print("✓ Agent created successfully!")
print("  The agent is now ready to provide planet information")
print("  All responses will follow the PlanetInfo structure")
print("-" * 60 + "\n")
```

```python
print("=" * 60)
print("STEP 6: Testing Structured Output")
print("=" * 60)
print("Asking the agent: 'Provide details about Mars.'")
print("The agent will return data in PlanetInfo format")
print("-" * 60 + "\n")

# Create the task message
task = TextMessage(content="Provide details about Mars.", source="user")

# Run the agent with the task
response = await agent.on_messages([task], cancellation_token=None)

# Extract and parse the structured data from response
import json
content_str = response.chat_message.content

print(f"Raw response type: {type(content_str)}")
print(f"Raw response: {content_str}\n")

# Parse the JSON string into PlanetInfo object
structured_data = PlanetInfo.model_validate_json(content_str)

print("=" * 60)
print("STRUCTURED OUTPUT RECEIVED:")
print("=" * 60)
print(f"Parsed object type: {type(structured_data)}")
print(f"Planet Name: {structured_data.name}")
print(f"Distance from Sun: {structured_data.distance_from_sun_km:,} km")
print(f"Supports Life: {structured_data.has_life}")
print("=" * 60)
print("\n✓ Success! The AI returned perfectly structured data!")
print("  The JSON was automatically parsed into a PlanetInfo object")
```

## 🎯 Key Benefits of Structured Output:

**Without Structured Output (Traditional Way):**
- ❌ AI returns free-form text: "Mars is 227.9 million km from the sun..."
- ❌ You need to parse the text manually
- ❌ Inconsistent format every time
- ❌ Error-prone extraction

**With Structured Output (What We Did):**
- ✅ AI returns a `PlanetInfo` object with exact fields
- ✅ Direct access: `structured_data.name`, `structured_data.distance_from_sun_km`
- ✅ Guaranteed consistent format
- ✅ Type-safe (name is always string, distance is always float, etc.)
