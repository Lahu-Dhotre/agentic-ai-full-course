# AI Agents: With Tools vs Without Tools

This notebook demonstrates two approaches:
1. **Agent WITHOUT Tools** - Relies only on the LLM's knowledge
2. **Agent WITH Tools** - Can use external functions (calculator) for accurate calculations

## Part 1: Import Libraries and Setup

```python
print("Importing required libraries...")
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_core.tools import FunctionTool
from dotenv import load_dotenv
from autogen_agentchat.messages import TextMessage
import os

# Load environment variables
load_dotenv()
openai_key = os.getenv("OPEN_AI_KEY")
print("✓ Libraries imported and API key loaded\n")
```

```python
# Create model client (same for both agents)
model_client = OpenAIChatCompletionClient(
    model="gpt-4", 
    api_key=openai_key
)
print("✓ Model client created with GPT-4\n")
```

## Part 2: Agent WITHOUT Tools

This agent only uses the LLM's knowledge. It cannot perform actual calculations.

```python
print("Creating agent WITHOUT tools...")
agent_without_tools = AssistantAgent(
    name="basic_assistant",
    model_client=model_client,
    system_message="You are a helpful AI assistant."
)
print("✓ Agent created (no tools)\n")
```

```python
print("Testing agent WITHOUT tools...")
print("Question: What is 456789 * 123456?\n")

response = await agent_without_tools.on_messages(
    [TextMessage(content="What is 456789 * 123456?", source="user")], 
    cancellation_token=None
)

print(f"Answer: {response.chat_message.content}\n")
print("Note: The agent might estimate or give an approximate answer since it can't use a calculator.")
```
